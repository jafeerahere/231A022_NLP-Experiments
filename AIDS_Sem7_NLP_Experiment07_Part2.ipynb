```python
print("Performed by Jafeera Ansari(UIN: 231A022)")

import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download('tagsets_json')
nltk.download('maxent_ne_chunker_tab')
nltk.download("maxent_ne_chunker")
nltk.download("words")
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022)")
from nltk import word_tokenize, pos_tag, ne_chunk
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022)")
text = "Emma Johnson traveled from London to Paris to attend an international technology conference organized by UNESCO. She carried a laptop, camera, and backpack while taking notes during the event. After the conference, she visited the Eiffel Tower and bought a souvenir from a nearby shop. Later, Emma met researchers from Google and Microsoft to discuss a new artificial intelligence project. Before returning home, she boarded a train, read a book, and enjoyed a cup of coffee during the journey."

#Tokenize the text
tokens = word_tokenize(text)
tokens
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022)")
tagged_tokens = pos_tag(tokens)
tagged_tokens
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022)")
#Perform named entity recognition
named_entities = ne_chunk(tagged_tokens)
named_entities
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022)")
import spacy

#Load the spaCy English model
nlp = spacy.load("en_core_web_sm")

#Process the text with spaCy
doc = nlp(text)
doc
```
