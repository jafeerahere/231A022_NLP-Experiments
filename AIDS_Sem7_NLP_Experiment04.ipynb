```python
text="I was working today"
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
print("Name:Ansari Jafeera , UIN: 231A022")
import string
print (string.punctuation)
```

```python
print("Name:Ansari Jafeera , UIN: 231A022")
text_lower=text.lower()
print(text_lower)
```

```python
print("Name:Ansari Jafeera , UIN: 231A022")

tokens=word_tokenize(text_lower)
print(tokens)
```

```python
print("Name:Ansari Jafeera , UIN: 231A022")
text_clean=[]
for word in tokens:
  if word not in string.punctuation:
    text_clean.append(word)
print(text_clean)
```

```python
print("Name:Ansari Jafeera , UIN: 231A022")
lemmatizer = WordNetLemmatizer()
lemmas=[lemmatizer.lemmatize(word) for word in text_clean]
print(lemmas)
```

```python

lemmatizer.lemmatize("working",pos="n")
```

```python
print("Name:Ansari Jafeera , UIN: 231A022")
[(token , lemma) for token, lemma in zip(tokens,lemmas)]
```
