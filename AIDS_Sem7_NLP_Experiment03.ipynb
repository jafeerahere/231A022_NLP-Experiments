```python
print("Jafeera Ansari", "UIN: 231A022\n")
import re
patter=r'https?://www\.\S+|www\.\S+|\S+\@\S+|\@\S+'
text_url=re.sub(patter," ",lower)
print(text_url)
```

```python
import emoji
text_emoji = emoji.replace_emoji(text_url, replace="")
print(text_emoji)
```

```python
from typing_extensions import Pattern
print("Jafeera Ansari", "UIN: 231A022\n")
patter =r'\d+\S+'
num=re.sub(patter," ",text_emoji)
print(num)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
print("Jafeera Ansari", "UIN: 231A022\n")
token=word_tokenize(num)
print(token)
```

```python
print("Jafeera Ansari", "UIN: 231A022\n")
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("Jafeera Ansari", "UIN: 231A022\n")
print("With punctutaion:", token)
print()
text_punch=[]
for word in token:
  if word not in punctuation:
    text_punch.append(word)
print ("Without punctuation:",text_punch)
```

Remove stop words


```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
print(stopwords.words('english'))
```

```python
print("Jafeera Ansari", "UIN: 231A022\n")
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
text_without_stopwords = [word for word in text_punch if word.lower() not in stop_words]
print("Without stopwords:", text_without_stopwords)
```

```python
text_clean = []
for word in text_without_stopwords:
  if word.lower() not in stop_words:
    text_clean.append(word)
print(text_clean)
```

stemming

```python
from nltk.stem import PorterStemmer
ps=PorterStemmer()
text_stem=[ps.stem(i) for i in text_clean]
print("Before Stemming:",text_clean)
print("After Stemming:",text_stem)
```

lemmitization

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lm=WordNetLemmatizer()
text_lemma=[lm.lemmatize(i) for i in text_clean]
print("stem text:",text_stem)
print("After Lemmatization:",text_lemma)
```
