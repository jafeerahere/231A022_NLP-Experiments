```python
print("Performed by Jafeera Ansari(UIN: 231A022)")
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('tagsets_json')
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
#sample Sentence
sentence = "I am Jafeera, chaotic human, I'll go watch football in 2030"
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
tokens = word_tokenize(sentence)
tokens
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
tagged_tokens = pos_tag(tokens)
tagged_tokens
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
nltk.help.upenn_tagset("CD")
```

```python
grammar = "NP: {<.*>+}"
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
#Create a chunk and parse the sentence
chunk_parser = RegexpParser(grammar)
chunked = chunk_parser.parse(tagged_tokens)
chunked
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
grammar1 = "VP: {<VB.*>+}"
chunked_parser = RegexpParser(grammar1)
chunked1 = chunked_parser.parse(tagged_tokens)
chunked1
```

```python
print("Performed by Jafeera Ansari(UIN: 231A022))")
import svgling
print(chunked1)
svgling.draw_tree(chunked1)
```
